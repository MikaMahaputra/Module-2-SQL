## Step 2.

Create a dictionary with 5 keys and each key has a value in the form of a list with 10 items. The contents of the dictionary follow the output of **Figure 1.**.

```python
data = {
    "Name":     ["Andi", "Budi", "Caca", "Dedi", "Erika", "Fahri", "Gina", "Hani", "Indro", "Jarot"],
    "Age":      [24, 25, 30, 32, 28, 24, 29, 31, 30, 29],
    "Gender":   ["Male", "Male", "Female", "Female", "Female", "Male", "Female", "Female", "Male", "Male"],
    "Location": ["Jakarta", "Tangerang Selatan", "Jakarta", "Bandung", "Jakarta", "Surabaya", "Bandung", "Yogyakarta", "Jakarta", "Tangerang Selatan"]
}
df = pd.DataFrame(data)
```

<br>

## Step 3.

Create a DataFrame using the data dictionary in **step 2** and store it in the variable df.

```python
print(df)
```

<br>

## Step 4.
Change the order of the columns to be as follows: Name, Gender, Age, Location, Job.

```python
df["Job"] = ["Data Analyst", "Data Engineer", "Data Scientist", "Data Scientist", "Data Analyst", "Data Engineer", "ML Engineer", "ML Researcher", "BI Analyst", "Data Analyst"]
df = df[["Name", "Gender", "Age", "Location", "Job"]]
print(df)
```

## Step 5. 
Add 1 new column named 'Hobby' and fill in the data as shown in the following output.

```python
df["Hobby"] = ["Gaming", "Futsal", "Reading", "Gaming", "Cooking", "Traveling", "Basketball", "Traveling", "Futsal", "Gaming"]
df = df[["Name", "Gender", "Age", "Location", "Job", "Hobby"]]
print(df)
```

<br>

## Step 6.
Use **.iloc** to retrieve data for Dedi, Fahri, Hani

```python
ilocfilter = df.iloc[[3, 5, 7]]
print(ilocfilter)

#Alternative Solution
cond = df["Name"].isin(values = ["Dedi", "Fahri", "Hani"])
df.loc[cond]
```

<br>

## Step 7.
Use **.loc** to display the name, location and job of **Caca** and **Indro**

```python
step7 = (df.loc[[2, 8], ["Name", "Location", "Job"]])
print(step7)
```

<br>

## Step 8.

Use **.loc** to display all data from **Hani**, **Indro**, **Jarot**

```python
step8 = df.loc[7:9]
print (step8)
```

<br>

## Step 9.
Who is over 30 years old?

```python
step9 = df['Age'] > 30
display(df.loc[step9,:])

#Alternative solution
display(df.loc[(df["Age"]>30), "Name"])
```

<br>

## Step 10.
Who is **under 30** years old and whose hobby is **gaming**?

```python
step10age = df ["Age"] < 30  
step10hobby = df ["Hobby"] == "Gaming"
df.loc[step10age | step10hobby]
```

<br>

## Step 11.
Show the name of the person whose job is Data Analyst! Show only the name column in Series form!

```python
step11 = df[df["Job"] == "Data Analyst"]["Name"]
print(step11)
```

<br>

## Step 12.
What are the jobs of people who live in Jakarta? Show their names and jobs! 

```python
step12 = df[df["Location"] == "Jakarta"][["Name", "Job"]]
print(step12)
```

<br>

## Step 13.
What are the hobbies of the women in the data? Show only the Name, Hobby, Age columns in order!

```python
step13 = df[df["Gender"] == "Female"][["Name", "Hobby", "Age"]]
print(step13)
```

<br>

## Step 14.
Who is older than Erika? Show only the name column in DataFrame!

```python
step14 = df[df["Age"] > 28]["Name"]
print(step14)
```

<br>

## Step 15.
Based on the data, are there any women whose hobby is traveling?

```python
result = df[(df["Gender"] == "Female") & (df["Hobby"] == "Traveling")].any().any()
print(result) 
```
