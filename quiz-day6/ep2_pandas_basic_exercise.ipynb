```python
import pandas as pd
import seaborn as sns
```

```python
df = sns.load_dataset('mpg')
df
```

## Task 1 - Tampilkan nilai yang paling sering muncul pada kolom 'origin'

```python
# Your code here
display(df["origin"].mode()[0])
display(df["origin"].count())
```

<br>

## Task 2 - Tampilkan rata-rata horsepower

```python
# Your code here
df["horsepower"].mean().round(2)
```

<br>

## Task 3 - Tampilkan rata-rata acceleration tiap cylinder

```python
#Your code here
df.groupby("cylinders")["acceleration"].mean()
```

<br>

## Task 4 - Tampilkan rata-rata acceleration tiap cylinder dan origin

```python
# Your code here
df.groupby(["cylinders", "origin"])["acceleration"].mean()
```

<br>

## Task 5 - Tampilkan rata-rata acceleration tiap cylinder dan origin, kemudian urutkan hasilnya dari acceleration paling tinggi.

```python
# Your code here
df.groupby(["cylinders", "origin"])["acceleration"].mean().sort_values(ascending=False, inplace=False)
```

<br>

## Task 6 - Tampilkan jumlah mobil tiap cylinder

```python
# Your code here
df.groupby(["cylinders"])["name"].count()
```

<br>

## Task 7 - Tampilkan mobil dari japan dengan nilai mpg paling tinggi

```python
# Your code here
car_japan = df[df["origin"] == "japan"]
car_japan.sort_values("mpg", ascending=False, inplace=False)
```

<br>

## Task 8 - Displacement adalah jumlah volume total dari seluruh cylinder. Buat kolom baru bernama 'volume_per_cylinder'.

```python
# Kode
df["volume_per_cylinder"] = df["displacement"] / df["cylinders"]
df.head()
```

<br>

## Task 9 - Tampilkan jumlah mobil keluaran tahun 1975

```python
model_75 = df['model_year'] == 75
display(df.loc[model_75])
display(df.loc[model_75, "name"].count())
```

<br>

## Task 10 - Buat kolom baru bernama 'origin_brand' berisi nama origin dan brand-nya.

Contoh:

-usa chevrolet
-usa buick
-europe vw

```python
# kode
df["origin_brand"] = df["origin"] + " - " + df["name"]
df
```

<br>

## Task 11 - Tambahkan kolom baru dengan nama kpl yang merupakan konversi dari mpg menjadi kilometer per liter (1 mpg = 0.42514 kpl).

```python
# kode
df["kpl"] = (df["mpg"] * 0.42514).round(2)
df
```
