## Step 2. Load and Explore Dataset

Load dataset into pandas

## Step 4. Remove unnecessary columns
Some columns, like **EmployeeCount**, **Over18**, and **EmployeeNumber**, might not add much value to the analysis. Remove them.

```python
df.drop(columns= ["EmployeeCount", "Over18", "EmployeeNumber"])
df
```

<br>

## Step 5. Create new columns

Create a column seniority_level based on YearsAtCompany:
- junior for employees with less than 3 years
- mid-level for employees with 3 to 6 years
- senior for employees with more than 6 years

```python
def seniority(years):
    if years < 3:
        return "junior"
    elif years <= 6:
        return "mid-level"
    else:
        return "senior"

df["seniority_level"] = df["YearsAtCompany"].apply(seniority)
df
```

<br>

## Step 6. Group by department and calculate average salary
Group the data by Department and calculate the average MonthlyIncome for each department

```python
df.groupby("Department")["MonthlyIncome"].mean()
```

<br>

## Step 7. Analyze attrition rates
1. Calculate the overall attrition rate (percentage of employees who have left).
1. Calculate the attrition rate by seniority_level and JobRole.

```python
# Overall attrition rate
attrition_rate = (df["Attrition"] == "Yes").sum() / len(df) * 100
print(f"Overall attrition rate: {attrition_rate:.2f}%")
```

```python
# Attrition rate by seniority level
df.groupby("seniority_level")["Attrition"].apply(lambda x: (x == "Yes").mean() * 100)
```

```python
# Attrition rate by job role
df.groupby("JobRole")["Attrition"].apply(lambda x: (x == "Yes").mean() * 100)
```

<br>

## Step 8. Filter the dataset

1. Filter the dataset to show only employees in the Sales department with more than 5 years at the company who have left (attrition = Yes)
1. Find all employees with a JobSatisfaction rating of 4 (highest) who still left the company.

```python
# Sales department employees with more than 5 years who left
df[(df["Department"] == "Sales") & (df["YearsAtCompany"] > 5) & (df["Attrition"] == "Yes")]
```

```python
# Employees with high job satisfaction who left
df[(df["JobSatisfaction"] == 4) & (df["Attrition"] == "Yes")]
```

<br>

## Step 9. Group by multiple columns

Group by JobRole and seniority_level and calculate the average MonthlyIncome.

```python
df.groupby(["JobRole", "seniority_level"])["MonthlyIncome"].mean()
```

<br>

## Step 10. Create a new column for income per year at the company
Create a new column income_per_year by dividing the employee's MonthlyIncome by the number of years at the company. Handle cases where YearsAtCompany is 0 (to avoid division by zero).

```python
df["income_per_year"] = df["MonthlyIncome"] / df["YearsAtCompany"].replace(0, np.nan, inplace=True)
df
```

<br>

## Step 11. Pivot tables for attrition analysis

1. Create a pivot table to analyze attrition by Department and JobSatisfaction.
1. reate another pivot table to analyze attrition by seniority_level and WorkLifeBalance.

```python
# Attrition by Department and JobSatisfaction
pd.pivot_table(df, 
               values="Attrition", 
               index="Department", 
               columns="JobSatisfaction", 
               aggfunc=lambda x: (x == "Yes").mean() * 100)
```

```python
# Attrition by Seniority level and WorkLifeBalance
pd.pivot_table(df, 
               values="Attrition", 
               index="seniority_level", 
               columns="WorkLifeBalance", 
               aggfunc=lambda x: (x == "Yes").mean() * 100)
```
