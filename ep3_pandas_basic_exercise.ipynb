**Audi**

**Hyundai**

```python
df_hyundai = pd.read_csv('hyundai.csv')
df_hyundai.head()
```

**Toyota**

```python
df_toyota = pd.read_csv('toyota.csv')
df_toyota.head()
```

## **No.1**

- Jika Anda perhatikan, nama kolom 'tax(£)' di dataset hyundai berbeda dengan di dataset lainnya.
- Ubah nama kolom tersebut menjadi 'tax' agar sama dengan di dataset yang lain.

```python
df_hyundai.rename(columns={"tax(£)": "tax"}, inplace=True)
df_hyundai
```

## **No.2**

- Buat kolom baru bernama 'brand' dengan isi nama brand mobilnya, letakkan pada kolom paling kiri.
- Misalnya:
    - Untuk data frame audi, isi kolom 'brand' dengan 'Audi' semua!
    - Untuk data frame hyundai, isi kolom 'brand' dengan 'Hyundai' semua!
    - Untuk data frame toyota, isi kolom 'brand' dengan 'Toyota'semua!


```python
df_audi["brand"] = pd.Series(["Audi" for x in range(len(df_audi.index))])
df_hyundai["brand"] = pd.Series(["Hyundai" for x in range(len(df_audi.index))])
df_toyota["brand"] = pd.Series(["Toyota" for x in range(len(df_audi.index))])
display(df_audi)
display(df_hyundai)
display(df_toyota)
```

## **No.3**

- Gabungkan ketiga data frame tersebut menjadi sebuah data frame bernama df_car. Atur agar indexnya tidak ada yang berulang.


```python
#df_car = df_audi.merge(df_hyundai, on='brand', how='inner').merge(df_toyota, on='brand', how='outer').dropna(axis=1)
df_car = pd.concat([df_audi, df_hyundai, df_toyota], axis=0, ignore_index=True)
df_car
```

## **No.4**

- Pada kolom 'model' di sebagian data, ada whitespace tambahan di awal atau di akhir yang tidak berguna.
- Hapus whitespace tersebut dari semua data pada kolom 'model'!

```python
df_car["model"].str.strip()
df_car
```

## **No.5**

- Dari df_car, tampilkan rata-rata pajak/tax berdasarkan brand dan transmission. Lalu urutkan dari tax yang paling tinggi.


```python
mediantax = df_car.groupby(["brand", "transmission"])["tax"].median().sort_values(ascending=False)
print(mediantax)
```

## **No.6**

- Anda diminta untuk mencari mobil Toyota Camry yang harganya (price) di bawah rata-rata harga Camry dan jarak tempuhnya (mileage) di bawah rata-rata mileage Camry juga.
- Tampilkan mobil yang sesuai dengan kriteria tersebut!

```python
camry = df_car[(df_car["brand"] == "Toyota") & (df_car["model"] == "Camry")]
avgprice = camry["price"].mean()
avgmiles = camry["mileage"].mean()
bestcamry = camry[(camry["price"] < avgprice) & (camry["mileage"] < avgmiles)]
print(bestcamry)
```

## **No.9**

Pada tahun berpa mobil Toyota paling banyak muncul


```python
dftoyota = df_car[df_car["brand"] == "Toyota"]
toyotafreq = dftoyota["year"].mode()[0]

print(toyotafreq)
```

## **No.10**

Tampilkan jumlah mobil totoya keluaran 2019 untuk tiap model


```python
toyota2019 = df_car[(df_car["brand"] == "Toyota") & (df_car["year"] == 2019)]
result = toyota2019.groupby("model")["brand"].count()
print(result)
```

## **No.11**

Tampilkan model dari Toyota yang paling irit (mpg paling tinggi)


```python
dftoyota = df_car[df_car["brand"] == "Toyota"]
mpgtoyota = dftoyota.loc[dftoyota["mpg"].idxmax(), "model"]
print(mpgtoyota)
```

## **No.12**

## **No.13**

Buat dataframe baru dengan pivot table berisi rata-rata tax tiap transmission dan fuelType

```python
pivot_tax = df_car.pivot_table(values="tax", index="transmission", columns="fuelType", aggfunc="mean")
print(pivot_tax)
```

## **No.14**

Ada berapa model yang hanya ada 1 unit mobil di dataset ini? (mobil langka)

```python
model_amt = df_car.groupby("model")["brand"].count()
rare_models = model_amt[model_amt == 1]
print("There are:",len(rare_models), "rare car models")
```
