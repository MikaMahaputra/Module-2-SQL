```python
df = pd.read_excel('melb.xlsx')
df.head()
```

No.1
- Tampilkan rumah paling mahal
- Tandai dengan icon pin berwarna merah
- Beri tooltip berupa "Rumah Termahal"
- Beri popup berupa alamat/address

```python
most_exp = df.loc[df['Price'].idxmax()]

m1 = folium.Map(location=[most_exp['Lattitude'], most_exp['Longtitude']], zoom_start=15)

marker = folium.Marker(
    location=[most_exp['Lattitude'], most_exp['Longtitude']],
    tooltip='Most Expensive',
    popup=folium.Popup(most_exp['Address'], max_width=200),
    icon=folium.Icon(color='red', icon='map-pin', prefix='fa')
)

marker.add_to(m1)
m1
```

No.2

- Ceritanya kita sedang mencari apartment (type='u') dengan 2 kamar tidur (Bedroom2) dengan harga yg murah
- Coba tampilkan peta kota Melbourne
- Berikan marker berupa pin pada 5 apartment termurah
- Berikan tooltip berupa harga rumah
- Berikan popup berupa Suburb dan Address

```python
apts = df[(df['Type'] == 'u') & (df['Bedroom2'] == 2)].nsmallest(5, 'Price')

m2 = folium.Map(location=[-37.8136, 144.9631], zoom_start=11)

for _, row in apts.iterrows():
    folium.Marker(
        location=[row['Lattitude'], row['Longtitude']],
        tooltip=f"Price: ${row['Price']:,.0f}",
        popup=folium.Popup(
            f"<b>{row['Suburb']}</b><br>{row['Address']}",
            max_width=200
        ),
        icon=folium.Icon(color='blue', icon='home', prefix='fa')
    ).add_to(m2)

m2
```
