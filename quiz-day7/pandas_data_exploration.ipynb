## **Dataset**

**Acknowledgements**<br>
Collected by Cai-Nicolas Ziegler in a 4-week crawl (August / September 2004) from the Book-Crossing community with kind permission from Ron Hornbaker, CTO of Humankind Systems. Contains 278,858 users (anonymized but with demographic information) providing 1,149,780 ratings (explicit / implicit) about 271,379 books.

**Content**<br>
The Book-Crossing dataset comprises 3 files.
- Users <br>
Contains the users. Note that user IDs (User-ID) have been anonymized and map to integers. Demographic data is provided (Location, Age) if available. Otherwise, these fields contain NULL-values.
- Books <br>
Books are identified by their respective ISBN. Invalid ISBNs have already been removed from the dataset. Moreover, some content-based information is given (Book-Title, Book-Author, Year-Of-Publication, Publisher), obtained from Amazon Web Services. Note that in case of several authors, only the first is provided. URLs linking to cover images are also given, appearing in three different flavours (Image-URL-S, Image-URL-M, Image-URL-L), i.e., small, medium, large. These URLs point to the Amazon web site.
- Ratings <br>
Contains the book rating information. Ratings (Book-Rating) are either explicit, expressed on a scale from 1-10 (higher values denoting higher appreciation), or implicit, expressed by 0.

## **Data Preparation**

### 1. Import all dataset (book, rating, and user)

```python
user = pd.read_csv("Users.csv")
rating = pd.read_csv("Ratings.csv")
book = pd.read_csv("Books.csv")
display(user, rating, book)
```

## **Exploratory Data Analysis (EDA)**

### 1. How many book are there

Number of books: ...

```python
num_books = book["ISBN"].nunique()
print("Number of books:", num_books)
```

### 2. How many author are there

Number of authors: ...

```python
num_authors = book["Book-Author"].nunique()
print("Number of authors:", num_authors)
```

### 3. Based on dataset, who is the author with the most number of books?

- Author: ...
- Number of books: ...

```python
best_author = book["Book-Author"].value_counts().idxmax()
best_book = book["Book-Title"].value_counts().max()
print("Top Author:", best_author)
print("Number of books:", best_book)
```

```python
top_publisher = book["Publisher"].value_counts().head(5)
print(top_publisher)
```

### 5. The book with the highest rating and most reviews? who is the author?

- Title: ...
- ISBN: ...
- Author: ...
- Publisher: ...
- Avg. Rating: ...
- Total Review: ...

```python
avg_rating = rating[rating["Book-Rating"] > 0].groupby("ISBN")["Book-Rating"].agg(["mean", "count"])
avg_rating = avg_rating[avg_rating["count"] >= 50].sort_values("mean", ascending=False)

top_isbn = avg_rating.index[0]
top_book = book[book["ISBN"] == top_isbn].iloc[0]

print("Title:", top_book["Book-Title"])
print("Author:", top_book["Book-Author"])
print("Publisher:", top_book["Publisher"])
print("Avg Rating:", round(avg_rating.loc[top_isbn, "mean"], 2))
print("Total Review:", avg_rating.loc[top_isbn, "count"])
```

### 6. Top 3 books with the highest avarage rating from `HarperFlamingo Canada (Publisher)`

1. title: ..., rating: ...
1. title: ..., rating: ...
1. title: ..., rating: ...

```python
hfc_isbns = book[book["Publisher"] == "HarperFlamingo Canada"]["ISBN"]
top3 = (rating[rating["ISBN"].isin(hfc_isbns) & (rating["Book-Rating"] > 0)]
        .groupby("ISBN")["Book-Rating"].mean()
        .sort_values(ascending=False)
        .head(3))

for isbn, avg in top3.items():
    title = book[book["ISBN"] == isbn]["Book-Title"].values[0]
    print(f"Title: {title}, Rating: {round(avg, 2)}")
```

### 7. Where do the most readers of `Edgehill Books (publisher)` come from ? (country)

1. Country: ...
1. Total Readers: ...

```python
eb_books = book[book["Publisher"] == "Edgehill Books"]["ISBN"]
eb_readers = rating[rating["ISBN"].isin(eb_books)]["User-ID"].unique()
eb_users = user[user["User-ID"].isin(eb_readers)]["Location"]
top_country = eb_users.str.split(",").str[-1].value_counts()

print(top_country)
```

### 8. Based on dataset, How many books were published in 2001 ?

```python
books_2001 = book[book["Year-Of-Publication"] == 2001]
print("Books published in 2001:", len(books_2001))
```

### 9. Average age of `Classical Mythology (ISBN: 0195153448)` readers

```python
classical_isbn = book[book["Book-Title"].str.contains("Classical Mythology")]["ISBN"]
readers = rating[rating["ISBN"].isin(classical_isbn)]["User-ID"]
avg_age = user[user["User-ID"].isin(readers)]["Age"].mean()
print("Average age:", round(avg_age, 2))
```

### 10. Find the 2 users who gave the highest average rating for each book from the publisher `Sellerio`

```python
sellerio_books = book[book["Publisher"] == "Sellerio"]["ISBN"]
sellerio_ratings = rating[rating["ISBN"].isin(sellerio_books) & (rating["Book-Rating"] > 0)]

top_users = (
    sellerio_ratings.groupby(["ISBN", "User-ID"])["Book-Rating"]
    .mean()
    .reset_index()
    .sort_values(["ISBN", "Book-Rating"], ascending=[True, False])
    .groupby("ISBN")
    .head(2)
)
print(top_users)
```
